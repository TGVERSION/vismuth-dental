```python
import json
import requests
```

```python
BASE_URL = "http://localhost:8001"


def print_response(response: requests.Response) -> None:
    print("Status:", response.status_code)
    try:
        data = response.json()
        print("JSON:")
        print(json.dumps(data, ensure_ascii=False, indent=2))
    except Exception:
        print("Text:")
        print(response.text)
```

```python
print("=== Health check ===")
health_resp = requests.get(f"{BASE_URL}/health", timeout=10)
print_response(health_resp)
```

# Autocomplete

### POST /autocomplete/complete

```python
complete_payload = {
    "field": "complaints",
    "prefix": "бол",
    "num_sequences": 5,
}
complete_resp = requests.post(
    f"{BASE_URL}/autocomplete/complete",
    json=complete_payload,
    timeout=30,
)
print_response(complete_resp)
```

```python
# Другой префикс и поле
complete_payload2 = {
    "field": "main_diagnosis",
    "prefix": "ост",
    "num_sequences": 3,
}
complete_resp2 = requests.post(
    f"{BASE_URL}/autocomplete/complete",
    json=complete_payload2,
    timeout=30,
)
print_response(complete_resp2)
```

```python
# Короткий префикс (< 2 символов) — сервис возвращает пустой список
short_prefix_resp = requests.post(
    f"{BASE_URL}/autocomplete/complete",
    json={"field": "complaints", "prefix": "а", "num_sequences": 5},
    timeout=10,
)
print_response(short_prefix_resp)
```
